**Public :** M1 Ingénierie • **Objectif :** visualiser l’effet des filtres de Gabor (orientation & fréquence spatiale) pour détecter des textures et motifs orientés.


```python
# === Setup: imports & image ===
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import os

img_path = 'lena_demo.png'
if not os.path.exists(img_path):
    raise FileNotFoundError("Place 'lena_demo.png' in the working directory.")
img_rgb = np.array(Image.open(img_path))
img_gray = np.dot(img_rgb[...,:3], [0.299, 0.587, 0.114]) / 255.0

plt.figure(figsize=(5,4))
plt.imshow(img_rgb)
plt.title("Lena image (provided)")
plt.axis('off')
plt.show()
```

```python
# === Generate Gabor kernel ===
def gabor_kernel(ksize=31, sigma=6.0, theta=np.pi/4, lam=10.0, gamma=0.5, phi=0.0):
    import numpy as np
    assert ksize % 2 == 1
    half = ksize // 2
    y, x = np.meshgrid(np.arange(-half, half+1), np.arange(-half, half+1), indexing='xy')
    x_p =  x*np.cos(theta) + y*np.sin(theta)
    y_p = -x*np.sin(theta) + y*np.cos(theta)
    gauss = np.exp(-(x_p**2 + (gamma**2)*(y_p**2)) / (2*sigma**2))
    sinus = np.cos(2*np.pi * x_p/lam + phi)
    g = gauss * sinus
    g -= g.mean()
    s = np.sqrt((g**2).sum())
    if s > 1e-8:
        g /= s
    return g
```

```python
# === Simple 2D convolution ===
def conv2_valid(img, kernel):
    H, W = img.shape
    k = kernel.shape[0]
    out_h = H - k + 1
    out_w = W - k + 1
    out = np.zeros((out_h, out_w), dtype=np.float32)
    kr = kernel[::-1, ::-1]
    for i in range(out_h):
        for j in range(out_w):
            out[i, j] = np.sum(img[i:i+k, j:j+k] * kr)
    return out

def conv2_same(img, kernel):
    k = kernel.shape[0]
    pad = k//2
    img_p = np.pad(img, ((pad,pad),(pad,pad)), mode='reflect')
    return conv2_valid(img_p, kernel)
```

```python
# === Orientation demo ===
thetas = [0, np.pi/6, np.pi/4, np.pi/3, np.pi/2]
k = 31
lam = 10.0
gamma = 0.5
sigma = 6.0

fig, axes = plt.subplots(1, len(thetas)+1, figsize=(14,3))
axes[0].imshow(img_gray, cmap='gray'); axes[0].set_title("Grayscale"); axes[0].axis('off')

for idx, th in enumerate(thetas, start=1):
    g = gabor_kernel(ksize=k, sigma=sigma, theta=th, lam=lam, gamma=gamma, phi=0.0)
    resp = conv2_same(img_gray, g)
    axes[idx].imshow(resp, cmap='gray')
    axes[idx].set_title(f"θ={th:.2f}")
    axes[idx].axis('off')

plt.suptitle("Response to Gabor filters (orientations)")
plt.tight_layout()
plt.show()
```

```python
# === Explore θ and λ ===
thetas = np.linspace(0, np.pi, 6, endpoint=False)
lambdas = [6.0, 10.0, 16.0]

fig, axes = plt.subplots(len(lambdas), len(thetas), figsize=(12,6))
for i, lam in enumerate(lambdas):
    for j, th in enumerate(thetas):
        g = gabor_kernel(ksize=31, sigma=6.0, theta=th, lam=lam, gamma=0.5, phi=0.0)
        resp = conv2_same(img_gray, g)
        axes[i, j].imshow(resp, cmap='gray')
        axes[i, j].set_xticks([]); axes[i, j].set_yticks([])
        axes[i, j].set_title(f"λ={lam:.0f}, θ={th:.2f}")
plt.suptitle("Filter bank (θ, λ)")
plt.tight_layout()
plt.show()
```

## Sharaine MALARVIJY
1. La réponse évolue de manière à avoir θ qui détermine l'orientation des contours détecté sur l'image.
2. L’effet de λ est la puissance de la determination de contours. λ petit efficace sur les détails fins et λ grand efficace sur les structures larges.
3. On normalise le noyau avant convolution car on veut se débarraser des composantes moyennes comme la luminance et que certaine caractéristique de l'image n'écrase pas les autres (on veut s'assurer que tous les filtres ont une amplitude comparable).

```python
def orientation_map_gabor(img, thetas, ksize=31, sigma=6.0, lam=20.0, gamma=0.5):
    H, W = img.shape
    responses = np.zeros((len(thetas), H, W), dtype=np.float32)

    for i, theta in enumerate(thetas):
        g = gabor_kernel(ksize=ksize, sigma=sigma, theta=theta, lam=lam, gamma=gamma)
        r = conv2_same(img, g)
        responses[i] = np.abs(r)

    idx_max = np.argmax(responses, axis=0)  # index du theta max par pixel
    orientation_map = np.array(thetas)[idx_max]
    magnitude_map = np.max(responses, axis=0)

    return orientation_map, magnitude_map

thetas = np.linspace(0, np.pi, 16, endpoint=False)  # 16 orientations
orientation_map, magnitude_map = orientation_map_gabor(img_gray, thetas)


plt.figure(figsize=(18,5))

plt.subplot(1,3,1)
plt.imshow(orientation_map, cmap='hsv')
plt.title("Carte d'orientation")
plt.colorbar(label='θ (radians)')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(magnitude_map, cmap='gray')
plt.title("Magnitude associée")
plt.axis('off')


plt.subplot(1,3,3)
hue = orientation_map / np.pi  # normalisation [0..1]

hsv_img = np.zeros((*orientation_map.shape, 3))
hsv_img[..., 0] = hue
hsv_img[..., 1] = 1.0
hsv_img[..., 2] = magnitude_map

from matplotlib.colors import hsv_to_rgb
rgb_img = hsv_to_rgb(hsv_img)

plt.imshow(rgb_img)
plt.title("Carte d'orientation colorée")
plt.axis('off')

plt.tight_layout()
plt.show()
```
